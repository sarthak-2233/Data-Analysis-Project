```python
def expected_delay_group(delay):
    if pd.isna(delay):
        return np.nan
    elif delay <= -16:
        return -2
    elif delay < 0:
        return -1
    elif delay < 15:
        return 0
    elif delay >= 180:
        return 12
    else:
        return int((delay - 15) // 15) + 1


dep_expected = df['DepDelay'].apply(expected_delay_group)
arr_expected = df['ArrDelay'].apply(expected_delay_group)

validation = pd.DataFrame({
    'Column': [
        'DepartureDelayGroups',
        'ArrivalDelayGroups'
    ],
    'Mismatches': [
        (
            df['DepartureDelayGroups'].notna() &
            (df['DepartureDelayGroups'] != dep_expected)
        ).sum(),

        (
            df['ArrivalDelayGroups'].notna() &
            (df['ArrivalDelayGroups'] != arr_expected)
        ).sum()
    ]
})

display(validation)
```

```python
def invalid_hhmm(series):
    values = series.dropna()
    
    hours = values // 100
    minutes = values % 100
    
    return ((hours > 23) | (minutes > 59)).sum()


time_validation = pd.DataFrame({
    'Column': time_cols,
    'Invalid HHMM Count': [
        invalid_hhmm(df[col])
        for col in time_cols
    ]
})

display(time_validation)
```

```python
#  1hr =60 min conversion
def hhmm_to_minutes(series):
    hours = series // 100
    minutes = series % 100
    return hours * 60 + minutes


time_minutes = pd.DataFrame({
    'DepTime': hhmm_to_minutes(df['DepTime']),
    'WheelsOff': hhmm_to_minutes(df['WheelsOff']),
    'WheelsOn': hhmm_to_minutes(df['WheelsOn']),
    'ArrTime': hhmm_to_minutes(df['ArrTime'])
})

display(time_minutes.head())
```

```python
# CONVERSION
chronological_times = time_minutes.copy()

# WheelsOff
chronological_times['WheelsOff'] = np.where(
    chronological_times['WheelsOff'] < chronological_times['DepTime'],
    chronological_times['WheelsOff'] + 1440,
    chronological_times['WheelsOff']
)

# WheelsOn
chronological_times['WheelsOn'] = np.where(
    chronological_times['WheelsOn'] < chronological_times['WheelsOff'],
    chronological_times['WheelsOn'] + 1440,
    chronological_times['WheelsOn']
)

# ArrTime
chronological_times['ArrTime'] = np.where(
    chronological_times['ArrTime'] < chronological_times['WheelsOn'],
    chronological_times['ArrTime'] + 1440,
    chronological_times['ArrTime']
)

display(chronological_times.head())
```

```python
chronology_check = pd.DataFrame({
    'Check': [
        'DepTime <= WheelsOff',
        'WheelsOff <= WheelsOn',
        'WheelsOn <= ArrTime'
    ],
    'Violations': [
        (
            chronological_times['DepTime'].notna() &
            chronological_times['WheelsOff'].notna() &
            (chronological_times['DepTime'] > chronological_times['WheelsOff'])
        ).sum(),

        (
            chronological_times['WheelsOff'].notna() &
            chronological_times['WheelsOn'].notna() &
            (chronological_times['WheelsOff'] > chronological_times['WheelsOn'])
        ).sum(),

        (
            chronological_times['WheelsOn'].notna() &
            chronological_times['ArrTime'].notna() &
            (chronological_times['WheelsOn'] > chronological_times['ArrTime'])
        ).sum()
    ]
})

display(chronology_check)
```

```python
## CHECKING INTEGRITY
exception_mask = (
    (
        chronological_times['WheelsOff'].notna() &
        chronological_times['WheelsOn'].notna() &
        (chronological_times['WheelsOff'] > chronological_times['WheelsOn'])
    )
    |
    (
        chronological_times['WheelsOn'].notna() &
        chronological_times['ArrTime'].notna() &
        (chronological_times['WheelsOn'] > chronological_times['ArrTime'])
    )
)

time_exceptions = df.loc[
    exception_mask,
    [
        'Year',
        
        'DayofMonth',
        'Flight_Number_Reporting_Airline',
        'Origin',
        'Dest',
        'DepTime',
        'WheelsOff',
        'WheelsOn',
        'ArrTime',
        'Cancelled',
        'Diverted',
        'CRSElapsedTime',
        'ActualElapsedTime',
        'AirTime',
        'TaxiOut',
        'TaxiIn'
    ]
].copy()

display(time_exceptions)
```
